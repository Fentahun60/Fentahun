```python
import pandas as pd
import numpy as np

# Example: your dataframe is df and the column is 'age_at_first_sex'

def recategorize_age(x):
    # Handle non-numeric values first
    if pd.isna(x):
        return np.nan
    if isinstance(x, str) and x.lower() == "not had sex":
        return "not had sex"

    # Convert to numeric if it's a string number
    try:
        x = float(x)
    except:
        return np.nan

    # Categorize
    if x < 15:
        return "<15"
    elif 15 <= x <= 17:
        return "15-17"
    elif 18 <= x <= 19:
        return "18-19"
    elif x >= 20:
        return ">20"
    else:
        return np.nan

# Apply to dataframe
df["age_at_first_sex"] = df["age_at_first_sex"].apply(recategorize_age)

# View result
df["age_at_first_sex"].value_counts(dropna=False)
```

```python
import pandas as pd

descriptive_columns = [
    'residence_type', 'highest_edu_level', 'sex_of_household_head',
    'freq_listen_radio', 'freq_watch_tv', 'owns_mobile_phone', 'wealth_index',
    'current_contraceptive_method', 'visited_health_facility_12mo', 'smokes_cigarettes',
    'distance_to_health_facility', 'waiting_time_concern_at_HF',
    'covered_by_health_insurance', 'breast_exam_by_provider', 'cervical_cancer_tested',
    'marital_status', 'age_at_first_sex', 'STI_last_12mo', 'ever_tested_hiv'
]

descriptive_table_data = []

# Ensure the 'weight' column exists in df_selected. It was previously created from df['v005'] / 1000000.
# If df_selected does not contain 'weight' from a previous step, uncomment the line below:
# df_selected['weight'] = df['v005'] / 1000000

for col in descriptive_columns:
    # Convert numeric columns that represent categories to string/object
    # to ensure they are treated as discrete categories for the descriptive table.
    if df_selected[col].dtype in ['int64', 'float64'] and col not in ['respondent_age', 'household_head_age', 'lifetime_sex_partners']:
        temp_series = df_selected[col].astype(str)
    else:
        temp_series = df_selected[col]

    # Group by the column (or temp_series for categorical treatment) and apply weights
    weighted_counts = df_selected.groupby(temp_series)['weight'].sum()
    total_weighted_sum = weighted_counts.sum()
    weighted_percentages = (weighted_counts / total_weighted_sum) * 100

    for category, count in weighted_counts.items():
        percent = weighted_percentages.get(category, 0)
        descriptive_table_data.append({
            'Variables': col,
            'Category': str(category),      # Ensure category is string for consistent output
            'Number': round(count, 0),       # Round count to 0 decimal places
            'Percent': round(percent, 0)     # Round percent to 0 decimal places
        })

# Create the final DataFrame
descriptive_table = pd.DataFrame(descriptive_table_data)

# Save to CSV file
csv_filename = "descriptive_table_weighted.csv"
descriptive_table.to_csv(csv_filename, index=False)
print(f"Table successfully saved to '{csv_filename}'\n")

print("Descriptive Table with Weighted Counts and Percentages:")
display(descriptive_table)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from math import sqrt

# Map 0/1 to No/Yes
y_mapped = df_selected["cervical_cancer_screened"].map({0: "No", 1: "Yes"})

# Counts
counts = y_mapped.value_counts()
n = counts.sum()

# Proportions
proportions = counts / n

# 95% CI (Normal approximation)
z = 1.96
ci_lower = proportions - z * np.sqrt((proportions * (1 - proportions)) / n)
ci_upper = proportions + z * np.sqrt((proportions * (1 - proportions)) / n)

# Create labels with CI
labels = [
    f"{label}\n{prop*100:.1f}% (95% CI: {low*100:.1f}–{up*100:.1f})"
    for label, prop, low, up in zip(
        proportions.index, proportions.values, ci_lower.values, ci_upper.values
    )
]

sizes = proportions.values
colors = ['#ff9999', '#66b3ff']
plt.figure(figsize=(7, 7))
plt.pie(
    sizes,
    labels=labels,
    startangle=140,
    colors=colors,
    wedgeprops={'edgecolor': 'black', 'linewidth': 1.2},
    textprops={'fontsize': 12}
)

plt.title("Proportion of Cervical Cancer Screening (95% CI)", fontsize=16)
plt.show()
```

```python
# ======================================================
# Full pipeline: Optuna + Final Models + Evaluation + ROC
# Revised with specified model list (LR, DT, RF, ET, XGBoost, CatBoost)
# ======================================================

import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ML libraries
import optuna
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.utils import resample

# Models
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import (
    RandomForestClassifier, ExtraTreesClassifier
)
from sklearn.linear_model import LogisticRegression

from xgboost import XGBClassifier
from catboost import CatBoostClassifier

# Metrics (added classification_report)
from sklearn.metrics import (
    roc_auc_score, average_precision_score,
    accuracy_score, precision_score, recall_score,
    f1_score, roc_curve, confusion_matrix,
    classification_report
)


# ======================================================
# 0. RFE-selected features
# ======================================================
selected_features = ranking_df[ranking_df["Selected"] == True]["Feature"].tolist()
print("Selected features (RFE):", selected_features)

X_train_sel = X_train_res[selected_features].copy()
X_test_sel  = X_test[selected_features].copy()

assert X_train_sel.shape[1] == len(selected_features)


# ======================================================
# Evaluation function
# ======================================================
def evaluate_model(name, model, X_test, y_test):
    try:
        probs = model.predict_proba(X_test)[:, 1]
    except Exception:
        try:
            scores = model.decision_function(X_test)
            probs = (scores - scores.min()) / (scores.max() - scores.min())
        except Exception:
            probs = model.predict(X_test)

    preds = (probs >= 0.5).astype(int)

    cm = confusion_matrix(y_test, preds)
    tn, fp, fn, tp = cm.ravel()

    specificity = tn / (tn + fp) if (tn + fp) > 0 else 0

    return {
        "model": name,
        "roc_auc": roc_auc_score(y_test, probs),
        "pr_auc": average_precision_score(y_test, probs),
        "accuracy": accuracy_score(y_test, preds),
        "precision": precision_score(y_test, preds, zero_division=0),
        "recall": recall_score(y_test, preds, zero_division=0),
        "f1": f1_score(y_test, preds, zero_division=0),
        "specificity": specificity,
        "confusion_matrix": cm,
        "probs": probs
    }

# Helper function to calculate bootstrap confidence intervals
def bootstrap_metric_ci(y_true, y_pred_proba, metric_name, n_bootstraps=500, alpha=0.05):
    """
    Calculates bootstrap confidence interval for a given metric.
    y_pred_proba should be probabilities for the positive class.
    metric_name: string identifier for the metric (e.g., "roc_auc", "accuracy")
    """
    n_samples = len(y_true)
    bootstrapped_scores = []

    # Convert y_true to numpy array for consistent indexing if it's a pandas Series
    y_true_np = y_true.to_numpy() if isinstance(y_true, pd.Series) else y_true

    # Convert y_pred_proba to numpy array if it's not already
    y_pred_proba_np = y_pred_proba.to_numpy() if isinstance(y_pred_proba, pd.Series) else y_pred_proba

    for _ in range(n_bootstraps):
        indices = resample(np.arange(n_samples), random_state=np.random.randint(0, 100000))
        y_true_bootstrap = y_true_np[indices]
        y_pred_proba_bootstrap = y_pred_proba_np[indices]

        # Ensure y_pred (0 or 1) for metrics that need it
        y_pred_bootstrap = (y_pred_proba_bootstrap >= 0.5).astype(int)

        score = None
        try:
            if metric_name == "roc_auc":
                # Check for single class in bootstrap sample which breaks AUC
                if len(np.unique(y_true_bootstrap)) < 2:
                    continue
                score = roc_auc_score(y_true_bootstrap, y_pred_proba_bootstrap)
            elif metric_name == "pr_auc":
                # Check for single class in bootstrap sample which breaks PR-AUC
                if len(np.unique(y_true_bootstrap)) < 2:
                    continue
                score = average_precision_score(y_true_bootstrap, y_pred_proba_bootstrap)
            elif metric_name == "accuracy":
                score = accuracy_score(y_true_bootstrap, y_pred_bootstrap)
            elif metric_name == "precision":
                score = precision_score(y_true_bootstrap, y_pred_bootstrap, zero_division=0)
            elif metric_name == "recall":
                score = recall_score(y_true_bootstrap, y_pred_bootstrap, zero_division=0)
            elif metric_name == "f1":
                score = f1_score(y_true_bootstrap, y_pred_bootstrap, zero_division=0)
            elif metric_name == "specificity":
                cm = confusion_matrix(y_true_bootstrap, y_pred_bootstrap)
                tn, fp, fn, tp = cm.ravel()
                score = tn / (tn + fp) if (tn + fp) > 0 else 0

            if score is not None:
                bootstrapped_scores.append(score)
        except Exception:
            # Silently skip errors during bootstrap
            pass

    bootstrapped_scores = np.array(bootstrapped_scores)

    # Filter out NaNs or infinities if any occurred
    bootstrapped_scores = bootstrapped_scores[np.isfinite(bootstrapped_scores)]

    if len(bootstrapped_scores) < (n_bootstraps * 0.9):  # If too many failed, return NaN
        return (np.nan, np.nan)

    # Calculate percentile-based CI
    lower = np.percentile(bootstrapped_scores, (alpha / 2) * 100)
    upper = np.percentile(bootstrapped_scores, (1 - alpha / 2) * 100)
    return (lower, upper)


# ======================================================
# 1. Optuna Objective Functions
# ======================================================

def optimize_decisiontree(trial):
    params = {
        "max_depth": trial.suggest_int("max_depth", 2, 30),
        "min_samples_split": trial.suggest_int("min_samples_split", 2, 10),
        "random_state": 42
    }
    model = DecisionTreeClassifier(**params)
    model.fit(X_train_sel, y_train_res)
    preds = model.predict_proba(X_test_sel)[:, 1]
    return roc_auc_score(y_test, preds)


def optimize_rf(trial):
    params = {
        "n_estimators": trial.suggest_int("n_estimators", 100, 600),
        "max_depth": trial.suggest_int("max_depth", 3, 30),
        "min_samples_split": trial.suggest_int("min_samples_split", 2, 10),
        "random_state": 42
    }
    model = RandomForestClassifier(**params, n_jobs=-1)
    model.fit(X_train_sel, y_train_res)
    preds = model.predict_proba(X_test_sel)[:, 1]
    return roc_auc_score(y_test, preds)


def optimize_extratrees(trial):
    params = {
        "n_estimators": trial.suggest_int("n_estimators", 100, 600),
        "max_depth": trial.suggest_int("max_depth", 3, 30),
        "random_state": 42
    }
    model = ExtraTreesClassifier(**params, n_jobs=-1)
    model.fit(X_train_sel, y_train_res)
    preds = model.predict_proba(X_test_sel)[:, 1]
    return roc_auc_score(y_test, preds)


def optimize_logistic(trial):
    C = trial.suggest_float("C", 1e-4, 100.0, log=True)
    model = Pipeline([
        ("scaler", StandardScaler()),
        ("lr", LogisticRegression(C=C, max_iter=5000, random_state=42))
    ])
    model.fit(X_train_sel, y_train_res)
    preds = model.predict_proba(X_test_sel)[:, 1]
    return roc_auc_score(y_test, preds)


def optimize_xgb(trial):
    params = {
        "n_estimators": trial.suggest_int("n_estimators", 100, 800),
        "max_depth": trial.suggest_int("max_depth", 3, 12),
        "learning_rate": trial.suggest_float("learning_rate", 0.01, 0.3),
        "subsample": trial.suggest_float("subsample", 0.5, 1.0),
        "colsample_bytree": trial.suggest_float("colsample_bytree", 0.4, 1.0),
        "gamma": trial.suggest_float("gamma", 0.0, 5.0),
        "eval_metric": "logloss",
        "random_state": 42
    }
    model = XGBClassifier(**params, n_jobs=-1)
    model.fit(X_train_sel, y_train_res)
    preds = model.predict_proba(X_test_sel)[:, 1]
    return roc_auc_score(y_test, preds)


def optimize_cat(trial):
    params = {
        "iterations": trial.suggest_int("iterations", 200, 1000),
        "depth": trial.suggest_int("depth", 3, 10),
        "learning_rate": trial.suggest_float("learning_rate", 0.01, 0.3),
        "random_state": 42,
        "verbose": 0
    }
    model = CatBoostClassifier(**params)
    model.fit(X_train_sel, y_train_res)
    preds = model.predict_proba(X_test_sel)[:, 1]
    return roc_auc_score(y_test, preds)


# ======================================================
# 2. Run Optuna Studies
# ======================================================
n_trials = 40

def run_search(func, name):
    print(f"\nStarting optimization for {name}...")
    optuna.logging.set_verbosity(optuna.logging.WARNING)
    study = optuna.create_study(direction="maximize")
    study.optimize(func, n_trials=n_trials, show_progress_bar=True)
    print(f"Best {name} ROC-AUC: {study.best_value:.4f}")
    print("Best Params:", study.best_params)
    return study.best_params


best_dt_params  = run_search(optimize_decisiontree, "DecisionTree")
best_rf_params  = run_search(optimize_rf, "RandomForest")
best_et_params  = run_search(optimize_extratrees, "ExtraTrees")
best_lr_params  = run_search(optimize_logistic, "LogisticRegression")
best_xgb_params = run_search(optimize_xgb, "XGBoost")
best_cat_params = run_search(optimize_cat, "CatBoost")


# ======================================================
# 3. Train final models
# ======================================================
final_models = {}

final_models["DecisionTree"] = DecisionTreeClassifier(
    **best_dt_params, random_state=42
).fit(X_train_sel, y_train_res)

final_models["RandomForest"] = RandomForestClassifier(
    **best_rf_params, random_state=42, n_jobs=-1
).fit(X_train_sel, y_train_res)

final_models["ExtraTrees"] = ExtraTreesClassifier(
    **best_et_params, random_state=42, n_jobs=-1
).fit(X_train_sel, y_train_res)

final_models["LogisticRegression"] = Pipeline([
    ("scaler", StandardScaler()),
    ("lr", LogisticRegression(C=best_lr_params["C"], max_iter=5000, random_state=42))
]).fit(X_train_sel, y_train_res)

final_models["XGBoost"] = XGBClassifier(
    **best_xgb_params, eval_metric="logloss", random_state=42, n_jobs=-1
).fit(X_train_sel, y_train_res)

cat_params = best_cat_params.copy()
cat_params.update({"random_state": 42, "verbose": 0})
final_models["CatBoost"] = CatBoostClassifier(**cat_params).fit(X_train_sel, y_train_res)


# ======================================================
# 4. Evaluate models + ROC curves
# ======================================================
results = []
plt.figure(figsize=(10, 8))

for name, model in final_models.items():
    res = evaluate_model(name, model, X_test_sel, y_test)
    results.append(res)
    fpr, tpr, _ = roc_curve(y_test, res["probs"])
    plt.plot(fpr, tpr, label=f"{name} (AUC={res['roc_auc']:.3f})")

plt.plot([0, 1], [0, 1], "k--", label="Random")
plt.title("ROC Curves", fontsize=14, fontweight="bold")
plt.xlabel("False Positive Rate", fontsize=12)
plt.ylabel("True Positive Rate", fontsize=12)
plt.legend(loc="lower right")
plt.grid(True, linestyle="--", alpha=0.6)
plt.tight_layout()
plt.show()


# ======================================================
# 5. Summary table + Top 2 confusion matrices + CIs
# ======================================================

# Augment results with bootstrap CIs
for r in results:
    y_true_test = y_test
    y_pred_proba_test = r["probs"]

    r["roc_auc_ci"]     = bootstrap_metric_ci(y_true_test, y_pred_proba_test, "roc_auc")
    r["pr_auc_ci"]      = bootstrap_metric_ci(y_true_test, y_pred_proba_test, "pr_auc")
    r["accuracy_ci"]    = bootstrap_metric_ci(y_true_test, y_pred_proba_test, "accuracy")
    r["precision_ci"]   = bootstrap_metric_ci(y_true_test, y_pred_proba_test, "precision")
    r["recall_ci"]      = bootstrap_metric_ci(y_true_test, y_pred_proba_test, "recall")
    r["specificity_ci"] = bootstrap_metric_ci(y_true_test, y_pred_proba_test, "specificity")
    r["f1_ci"]          = bootstrap_metric_ci(y_true_test, y_pred_proba_test, "f1")

# Create temporary DataFrame for sorting
summary_temp = pd.DataFrame([{
    "Model": r["model"],
    "ROC_AUC_val": r["roc_auc"],
    "ROC_AUC_ci": r["roc_auc_ci"],
    "PR_AUC_val": r["pr_auc"],
    "PR_AUC_ci": r["pr_auc_ci"],
    "Accuracy_val": r["accuracy"],
    "Accuracy_ci": r["accuracy_ci"],
    "Precision_val": r["precision"],
    "Precision_ci": r["precision_ci"],
    "Recall_val": r["recall"],
    "Recall_ci": r["recall_ci"],
    "Specificity_val": r["specificity"],
    "Specificity_ci": r["specificity_ci"],
    "F1_val": r["f1"],
    "F1_ci": r["f1_ci"]
} for r in results]).sort_values("ROC_AUC_val", ascending=False).reset_index(drop=True)

# Format final summary table with CI ranges
summary_formatted_data = []
for idx, row in summary_temp.iterrows():
    formatted_row = {
        "Model": row["Model"],
        "ROC_AUC": f"{row['ROC_AUC_val']:.3f} ({row['ROC_AUC_ci'][0]:.3f}-{row['ROC_AUC_ci'][1]:.3f})",
        "PR_AUC": f"{row['PR_AUC_val']:.3f} ({row['PR_AUC_ci'][0]:.3f}-{row['PR_AUC_ci'][1]:.3f})",
        "Accuracy": f"{row['Accuracy_val']:.3f} ({row['Accuracy_ci'][0]:.3f}-{row['Accuracy_ci'][1]:.3f})",
        "Precision": f"{row['Precision_val']:.3f} ({row['Precision_ci'][0]:.3f}-{row['Precision_ci'][1]:.3f})",
        "Recall": f"{row['Recall_val']:.3f} ({row['Recall_ci'][0]:.3f}-{row['Recall_ci'][1]:.3f})",
        "Specificity": f"{row['Specificity_val']:.3f} ({row['Specificity_ci'][0]:.3f}-{row['Specificity_ci'][1]:.3f})",
        "F1": f"{row['F1_val']:.3f} ({row['F1_ci'][0]:.3f}-{row['F1_ci'][1]:.3f})"
    }
    summary_formatted_data.append(formatted_row)

summary = pd.DataFrame(summary_formatted_data)

print("\nModel Performance Summary (with 95% CI):")
print(summary.to_string())

top_models = summary_temp["Model"].head(2).tolist()

for tm in top_models:
    print(f"\n--- {tm} ---")
    model_obj = final_models[tm]
    try:
        probs = model_obj.predict_proba(X_test_sel)[:, 1]
        preds = (probs >= 0.5).astype(int)
    except Exception:
        preds = model_obj.predict(X_test_sel)

    print("Confusion Matrix:\n", confusion_matrix(y_test, preds))
    print("Classification Report:\n", classification_report(y_test, preds))

# Save detailed numerical summary with lower/upper CIs to CSV
summary_numeric_for_csv = pd.DataFrame([{
    "Model": r["model"],
    "ROC_AUC": r["roc_auc"],
    "ROC_AUC_lower_ci": r["roc_auc_ci"][0],
    "ROC_AUC_upper_ci": r["roc_auc_ci"][1],
    "PR_AUC": r["pr_auc"],
    "PR_AUC_lower_ci": r["pr_auc_ci"][0],
    "PR_AUC_upper_ci": r["pr_auc_ci"][1],
    "Accuracy": r["accuracy"],
    "Accuracy_lower_ci": r["accuracy_ci"][0],
    "Accuracy_upper_ci": r["accuracy_ci"][1],
    "Precision": r["precision"],
    "Precision_lower_ci": r["precision_ci"][0],
    "Precision_upper_ci": r["precision_ci"][1],
    "Recall": r["recall"],
    "Recall_lower_ci": r["recall_ci"][0],
    "Recall_upper_ci": r["recall_ci"][1],
    "Specificity": r["specificity"],
    "Specificity_lower_ci": r["specificity_ci"][0],
    "Specificity_upper_ci": r["specificity_ci"][1],
    "F1": r["f1"],
    "F1_lower_ci": r["f1_ci"][0],
    "F1_upper_ci": r["f1_ci"][1]
} for r in results]).sort_values("ROC_AUC", ascending=False).reset_index(drop=True)

summary_numeric_for_csv.to_csv("model_performance_summary_with_ci.csv", index=False)
print("\nSaved detailed summary with numerical CIs to model_performance_summary_with_ci.csv")
```

```python
import pandas as pd
from sklearn.metrics import brier_score_loss
from sklearn.linear_model import LogisticRegression
import numpy as np
import matplotlib.pyplot as plt
from sklearn.calibration import calibration_curve

print("\n--- Additional Model Calibration Metrics ---")

calibration_metrics = []

# Loop through the existing results for each model
for r in results:
    model_name = r["model"]
    y_true_test = y_test
    y_pred_proba_test = r["probs"]

    # Calculate Brier Score
    brier_score = brier_score_loss(y_true_test, y_pred_proba_test)

    # Calculate Calibration Intercept
    # Fit a logistic regression model on predicted probabilities vs. true labels
    # The intercept of this model can indicate calibration (ideally close to 0)
    try:
        # Reshape for LogisticRegression, which expects 2D array for X
        lr_cal = LogisticRegression(penalty=None, solver='liblinear', random_state=42)
        lr_cal.fit(y_pred_proba_test.reshape(-1, 1), y_true_test)
        calibration_intercept = lr_cal.intercept_[0]
    except Exception as e:
        calibration_intercept = np.nan # In case fitting fails
        # Suppress repeated warnings if the solver issue is consistent
        # print(f"Warning: Could not calculate calibration intercept for {model_name}. Error: {e}")

    calibration_metrics.append({
        "Model": model_name,
        "Brier Score": f"{brier_score:.3f}",
        "Calibration Intercept": f"{calibration_intercept:.3f}"
    })

# Create a DataFrame for display
calibration_df = pd.DataFrame(calibration_metrics)

# Display the table, sorted by Brier Score (lower is better)
calibration_df = calibration_df.sort_values(by="Brier Score", ascending=True).reset_index(drop=True)
print(calibration_df.to_string())

print("\n--- Calibration Plots ---")
plt.figure(figsize=(10, 8))
ax = plt.gca()

for r in results:
    model_name = r["model"]
    if model_name == "CatBoost": # Only plot for CatBoost
        y_true_test = y_test
        y_pred_proba_test = r["probs"]
        brier_score = brier_score_loss(y_true_test, y_pred_proba_test)

        # Generate calibration curve data
        fraction_of_positives, mean_predicted_value = calibration_curve(
            y_true_test, y_pred_proba_test, n_bins=10
        )

        ax.plot(
            mean_predicted_value, fraction_of_positives, "s-",
            label=f"{model_name} (Brier: {brier_score:.3f})"
        )

ax.plot([0, 1], [0, 1], "k:", label="Perfectly calibrated")
ax.set_xlabel("Mean Predicted Value")
ax.set_ylabel("Fraction of Positives")
ax.set_title("Calibration Plot for CatBoost Model with Brier Score")
ax.legend(loc="lower right")
ax.grid(True)
plt.tight_layout()
plt.savefig("catboost_calibration_plot.png", dpi=300)
plt.show()
```

```python
# ============================================================
# CATBOOST SHAP ANALYSIS
# ============================================================

import shap
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd


# ============================================================
# 1. SELECT CATBOOST MODEL
# ============================================================

best_model_name = "CatBoost"

# Retrieve the already fitted CatBoost model
best_model = final_models[best_model_name]

print("=" * 70)
print(f"SHAP ANALYSIS FOR: {best_model_name}")
print("=" * 70)


# ============================================================
# 2. ENSURE X_TEST_SEL IS A DATAFRAME
# ============================================================

if not isinstance(X_test_sel, pd.DataFrame):
    X_test_sel = pd.DataFrame(
        X_test_sel,
        columns=selected_features
    )

print(f"\nTest dataset shape: {X_test_sel.shape}")
print(f"Number of features: {X_test_sel.shape[1]}")


# ============================================================
# 3. CREATE SHAP EXPLAINER
# ============================================================

print("\nCreating SHAP TreeExplainer...")

explainer = shap.TreeExplainer(best_model)


# ============================================================
# 4. CALCULATE SHAP VALUES
# ============================================================

print("Calculating SHAP values...")

shap_values = explainer.shap_values(X_test_sel)


# ============================================================
# 5. EXTRACT POSITIVE-CLASS SHAP VALUES
# ============================================================

if isinstance(shap_values, list):

    # Older SHAP versions:
    # [class 0 SHAP values, class 1 SHAP values]
    shap_values_for_plot = shap_values[1]

elif isinstance(shap_values, np.ndarray) and shap_values.ndim == 3:

    # Newer SHAP versions:
    # samples × features × classes
    shap_values_for_plot = shap_values[:, :, 1]

else:

    # Already a 2-dimensional array
    shap_values_for_plot = shap_values


# Convert to numpy array
shap_values_for_plot = np.asarray(shap_values_for_plot)


print(
    "\nSHAP values shape:",
    shap_values_for_plot.shape
)


# ============================================================
# 6. CHECK DIMENSIONS
# ============================================================

if shap_values_for_plot.shape[1] != X_test_sel.shape[1]:

    raise ValueError(
        f"Mismatch between SHAP values "
        f"({shap_values_for_plot.shape[1]} features) "
        f"and X_test_sel "
        f"({X_test_sel.shape[1]} features)."
    )


# ============================================================
# 7. CALCULATE MEAN ABSOLUTE SHAP VALUES
# ============================================================

mean_abs_shap = np.abs(
    shap_values_for_plot
).mean(axis=0)


# ============================================================
# 8. CREATE SHAP IMPORTANCE DATAFRAME
# ============================================================

shap_importance = pd.DataFrame({

    "Feature": X_test_sel.columns,

    "Mean_Absolute_SHAP": mean_abs_shap

})


# Sort from lowest to highest
# This makes the highest feature appear at the top
shap_importance = shap_importance.sort_values(
    "Mean_Absolute_SHAP",
    ascending=True
)


# Display SHAP importance table
print("\n" + "=" * 70)
print("SHAP FEATURE IMPORTANCE")
print("=" * 70)

print(
    shap_importance
    .sort_values(
        "Mean_Absolute_SHAP",
        ascending=False
    )
    .to_string(index=False)
)


# ============================================================
# 9. SHAP BAR PLOT WITH VALUES ON BARS
# ============================================================

print("\nGenerating SHAP bar plot...")


fig, ax = plt.subplots(
    figsize=(11, 8)
)


bars = ax.barh(

    shap_importance["Feature"],

    shap_importance["Mean_Absolute_SHAP"]

)


# ------------------------------------------------------------
# Add numerical SHAP values at the end of each bar
# ------------------------------------------------------------

max_shap = shap_importance[
    "Mean_Absolute_SHAP"
].max()


for bar, value in zip(
    bars,
    shap_importance["Mean_Absolute_SHAP"]
):

    ax.text(

        bar.get_width()
        + (max_shap * 0.01),

        bar.get_y()
        + bar.get_height() / 2,

        f"{value:.4f}",

        va="center",

        fontsize=9

    )


# ------------------------------------------------------------
# Axis labels and title
# ------------------------------------------------------------

ax.set_xlabel(
    "Mean Absolute SHAP Value",
    fontsize=11
)

ax.set_ylabel(
    "Features",
    fontsize=11
)

ax.set_title(
    "SHAP Feature Importance for CatBoost",
    fontsize=14,
    fontweight="bold"
)


# ------------------------------------------------------------
# Clean appearance
# ------------------------------------------------------------

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)


# Add a little space on the right for the numbers
ax.set_xlim(
    0,
    max_shap * 1.15
)


plt.tight_layout()
plt.show()


# ============================================================
# 10. SHAP BEESWARM SUMMARY PLOT
# ============================================================

print("\nGenerating SHAP beeswarm summary plot...")


plt.figure(
    figsize=(11, 8)
)


shap.summary_plot(

    shap_values_for_plot,

    X_test_sel,

    show=False

)


plt.title(
    "SHAP Summary Plot for CatBoost",
    fontsize=14,
    fontweight="bold"
)


plt.tight_layout()
plt.show()


# ============================================================
# 11. SHAP WATERFALL PLOT
# ============================================================

print("\nGenerating SHAP waterfall plot...")


# Select one observation from the test set
instance_index = 0


# Extract observation
instance_data = X_test_sel.iloc[
    instance_index
]


# Extract SHAP values for this observation
instance_shap_values = (
    shap_values_for_plot[
        instance_index
    ]
)


# ============================================================
# 12. DETERMINE BASE VALUE
# ============================================================

expected_value = explainer.expected_value


if isinstance(
    expected_value,
    (list, np.ndarray)
):

    expected_value = np.asarray(
        expected_value
    )

    if expected_value.ndim > 0:

        if expected_value.size > 1:

            # Positive class
            expected_value = expected_value[1]

        else:

            expected_value = expected_value.item()


# Convert to scalar
expected_value = float(
    np.asarray(
        expected_value
    ).reshape(-1)[0]
)


# ============================================================
# 13. CREATE SHAP EXPLANATION OBJECT
# ============================================================

shap_explanation = shap.Explanation(

    values=instance_shap_values,

    base_values=expected_value,

    data=instance_data.values,

    feature_names=X_test_sel.columns.tolist()

)


# ============================================================
# 14. WATERFALL PLOT
# ============================================================

shap.plots.waterfall(

    shap_explanation,

    max_display=15,

    show=False

)


plt.title(

    f"SHAP Waterfall Plot for "
    f"CatBoost (Test Instance {instance_index})",

    fontsize=14,

    fontweight="bold"

)


plt.tight_layout()
plt.show()


# ============================================================
# 15. SAVE SHAP IMPORTANCE TABLE
# ============================================================

shap_importance_output = (
    shap_importance
    .sort_values(
        "Mean_Absolute_SHAP",
        ascending=False
    )
    .reset_index(drop=True)
)


print("\n" + "=" * 70)
print("FINAL SHAP RANKING")
print("=" * 70)

print(
    shap_importance_output.to_string(
        index=False
    )
)


# ============================================================
# END OF SHAP ANALYSIS
# ============================================================
```
